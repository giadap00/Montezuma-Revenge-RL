```python
from google.colab import files
import numpy as np

rewards = np.load("rewards.npy")
intrinsic_rewards = np.load("intrinsic_rewards.npy")
extrinsic_rewards = np.load("extrinsic_rewards.npy")
loss = np.load("a2c_loss.npy")
icm_loss = np.load("icm_loss.npy")
steps = np.load("frames_per_reward.npy")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(30, 6))
plt.plot(rewards, label="Rewards")
plt.xlabel("Episodes")
plt.ylabel("Rewards")
plt.title("Training Rewards")
plt.legend()

plt.xticks(ticks=range(0, len(rewards), 100))

plt.show()
```

```python
print("Average of the rewards:", np.mean(rewards))

print("Maximum of the rewards:", np.max(rewards))
```
